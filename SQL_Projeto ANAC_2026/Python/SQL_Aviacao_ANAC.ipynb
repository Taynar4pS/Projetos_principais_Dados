**Projeto de SQL + PYTHON com base de dados da ANAC**

```python
import pandas as pd
```

```python
aviacao_janeiro = "VRA_2026_01.csv"
```

```python
# Sistemas do governo brasileiro frequentemente usam ';' em vez de ','
# e a codificação 'latin1' para lidar com acentos.
df_amostra = pd.read_csv(aviacao_janeiro, sep=';', encoding='utf-8-sig', nrows=100)

# 2. Padronização: transforma "Sigla ICAO Empresa Aérea" em "sigla_icao_empresa_aerea"
# Removemos acentos, trocamos espaços por 'underline' e deixamos tudo minúsculo.
df_amostra.columns = (
    df_amostra.columns
    .str.normalize('NFKD')
    .str.encode('ascii', errors='ignore')
    .str.decode('utf-8')
    .str.replace(' ', '_')
    .str.lower()
)

# 3. Visualizar as 5 primeiras linhas
print("\nPrimeiros registros:")
print(df_amostra.head())
```
