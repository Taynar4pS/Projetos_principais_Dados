Os campos do nosso dataframe são:

# 1 - Traga a média e a mediana dos preços - coluna Preco_Normal - por categoria de produto.
# Identifique as categorias que parecem ter um valor de média abaixo ou acima da mediana.

```python
#Calculo da média geral e por categoria
media_geral = df['Preco_Normal'].mean()
print('Média Geral:',media_geral)


media_categoria = df.groupby('Categoria')['Preco_Normal'].mean()
print('Média por categoria:', media_categoria)

```

```python
#Calculo da mediana geral e por categoria:
mediana_geral = df['Preco_Normal'].median()
print('Mediana Geral:', mediana_geral)

mediana_categoria = df.groupby('Categoria')['Preco_Normal'].median()
print('Mediana por categoria:', mediana_categoria)
```

Observa-se que, na maioria das categorias, a média dos preços é superior à mediana, indicando a presença de produtos com valores elevados que influenciam o preço médio e sugerem uma distribuição assimétrica dos preços.

Categorias do Data Frame:
* belleza-y-cuidado-personal
* comidas-preparadas
* congelados
* frutas
* instantaneos-y-sopas
* lacteos
* verduras


# 2 - Traga o desvio padrão por categoria de produto.
# Qual o comportamento da média e mediana nas categorias com maior desvio?

```python
#Calculo do desvio padrão
desvio_padrao_categoria= df.groupby('Categoria')['Preco_Normal'].std()
print('desvio padrão por categoria:',desvio_padrao_categoria)
```

A categoria lácteos apresenta o maior desvio padrão de preços, indicando elevada variabilidade e sugerindo a presença de produtos com valores bastante distintos dentro da mesma categoria. Em contraste, categorias como verduras e instantâneos e sopas apresentam menor dispersão, indicando preços mais homogêneos.

# 4 - Plote um gráfico de barras onde temos a média de descontos por categoria.

```python
# Seu código aqui
media_categoria_desconto = df.groupby('Categoria')['Preco_Desconto'].mean().reset_index()
print(media_categoria)
fig = px.bar(media_categoria_desconto, x='Categoria', y='Preco_Desconto')
fig.show()
```

* A análise da média de desconto por categoria indica que comidas preparadas apresenta os maiores valores médios de desconto, seguida por lácteos e congelados. Em contrapartida, a categoria instantâneos e sopas apresenta a menor média de desconto, sugerindo uma política promocional menos agressiva nesse segmento.
* Categorias com maior média de desconto tendem a concentrar produtos de maior valor agregado ou com maior necessidade de estímulo à venda, enquanto categorias com menor desconto médio podem apresentar maior estabilidade de demanda.
* Esse resultado é consistente com a análise de dispersão de preços, uma vez que categorias como lácteos e congelados também apresentaram maior variabilidade de preços.

# 5 - Plote um gráfico de mapa interativo agrupando os dados por categoria, marca e trazendo a média de desconto.

```python
# Seu código aqui
media_precos_desconto = df.groupby(['Categoria','Marca'])['Preco_Desconto'].mean().unstack()
print(media_precos_desconto)
fig = px.imshow(media_precos_desconto, text_auto=True, aspect="auto")
fig.show()
```

* A análise da média de descontos por categoria e marca evidencia um padrão esparso, no qual a maioria das marcas atua em apenas uma ou poucas categorias. Observa-se que os descontos concentram-se principalmente nas categorias de lácteos, beleza e cuidado pessoal e congelados, enquanto diversas combinações categoria–marca não apresentam descontos aplicados. Esse comportamento sugere uma estratégia de descontos direcionada, possivelmente focada em categorias específicas ou marcas com maior competitividade.
* Embora o heatmap permita identificar padrões gerais de concentração de descontos, a elevada quantidade de valores ausentes e zeros indica que análises mais focadas, como gráficos por categoria ou por marcas específicas, seriam mais adequadas para comunicação dos resultados
